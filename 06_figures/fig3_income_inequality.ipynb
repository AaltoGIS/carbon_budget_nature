```python
city_file_map = {
    "Helsinki": "./data/user_pois_pt_1.parquet",
    "Tampere":  "./data/user_pois_pt_1_tampere.parquet",
    "Turku":    "./data/user_pois_pt_1_turku.parquet",
    "Oulu":     "./data/user_pois_pt_1_oulu.parquet",
}

home_frames = []
for city, path in city_file_map.items():
    df = pd.read_parquet(path)
    df["city"] = city
    home_frames.append(df)

users_all = pd.concat(home_frames, ignore_index=True)

user_home_hex = (
    users_all[users_all["is_home"] == 1]
    .drop_duplicates(subset="user_id")
    [["user_id", "city", "home_gid9"]]
    .rename(columns={"home_gid9": "home_h3"})
)

```

```python
user_home_hex["lat"], user_home_hex["lon"] = zip(
    *user_home_hex["home_h3"].apply(h3.h3_to_geo)
)
```

```python
gdf_home = gpd.GeoDataFrame(
    user_home_hex,
    geometry=gpd.points_from_xy(user_home_hex["lon"], user_home_hex["lat"]),
    crs="EPSG:4326"
).to_crs("EPSG:3067")
```

```python

rttk = gpd.read_file("data/grid_data/rttk250m_tilv2019.shp")
```

```python
rttk_income = rttk[["grid_id", "hr_mtu", "geometry"]].copy()

gdf_home_income = gpd.sjoin(
    gdf_home,
    rttk_income,
    how="left",
    predicate="within"
)

# check how many failed to match (should be ~0 for points on land)
n_unmatched = gdf_home_income["hr_mtu"].isna().sum()
print(f"{n_unmatched} of {len(gdf_home_income)} home points unmatched")
```

```python
# Split into matched and unmatched
matched = gdf_home_income[gdf_home_income["hr_mtu"].notna()].copy()
unmatched = gdf_home_income[gdf_home_income["hr_mtu"].isna()].copy()

# Drop the failed join columns before retrying, keep original home geometry
unmatched = unmatched[gdf_home.columns].copy()  # back to original columns + geometry

# Nearest join for the unmatched subset only, with a max search distance
unmatched_recovered = gpd.sjoin_nearest(
    unmatched,
    rttk_income,
    how="left",
    max_distance=50,  # metres
    distance_col="dist_to_grid_m"
)

# check how many got recovered
n_recovered = unmatched_recovered["hr_mtu"].notna().sum()
print(f"{n_recovered} of {len(unmatched)} unmatched points recovered within 100m")

# still-unmatched after the 100m nearest attempt
still_unmatched = unmatched_recovered[unmatched_recovered["hr_mtu"].isna()]
print(f"{len(still_unmatched)} points remain unmatched — will be dropped")

# recombine: matched (within) + recovered (nearest ≤100m)
matched["dist_to_grid_m"] = 0.0  # exact containment, distance is 0 by definition
gdf_home_income_final = pd.concat(
    [matched, unmatched_recovered[unmatched_recovered["hr_mtu"].notna()]],
    ignore_index=True
)

print(f"Final matched: {len(gdf_home_income_final)} of {len(gdf_home)} home points "
      f"({len(gdf_home_income_final)/len(gdf_home)*100:.1f}%)")
```

```python
# Tilastokeskus uses two suppression mechanisms for hr_mtu:
# (1) NaN for cells nulled outright (small population)
# (2) -1 as a flag code for unreliable/withheld values
# Drop both before computing deciles.
n_before = len(gdf_home_income_final)
gdf_home_income_final = gdf_home_income_final.dropna(subset=["hr_mtu"])
gdf_home_income_final = gdf_home_income_final[gdf_home_income_final["hr_mtu"] >= 0]
n_after = len(gdf_home_income_final)
print(f"{n_before - n_after} points dropped (NaN or -1 suppression code)")
print(f"Final usable sample: {n_after} users ({n_after/len(gdf_home)*100:.1f}% of original)")

# Income decile, computed within the user population:
# do this globally (across all 4 cities pooled) or per-city.
# A global decile split allows direct comparison of deciles across cities.
gdf_home_income_final["income_decile"] = pd.qcut(
    gdf_home_income_final["hr_mtu"], 10, labels=False, duplicates="drop"
) + 1
print(gdf_home_income_final.groupby("income_decile")["hr_mtu"].agg(["min", "max", "count"]))
```

```python
gdf_home_income_final = gdf_home_income_final.merge(
    user_city_lookup[["user_id", "city"]],
    on="user_id",
    how="left",
    suffixes=("", "_lookup")
)


# check for unmatched (users with home/income data but no match in tour data, or vice versa)
print(gdf_home_income_final["city"].isna().sum(), "users unmatched to a city")

gdf_home_income_final = gdf_home_income_final.dropna(subset=["city"])

gdf_home_income_final["income_decile"] = (
    gdf_home_income_final.groupby("city")["hr_mtu"]
    .transform(lambda x: pd.qcut(x, 10, labels=False, duplicates="drop") + 1)
)

print(gdf_home_income_final.groupby(["city", "income_decile"])["hr_mtu"].agg(["min", "max", "count"]))
```

```python
decile_lookup = gdf_home_income_final[["user_id", "city", "income_decile"]]

compliance_rows = []
for city in cities:
    for mode in modes:
        df = data_dict[(city, mode)].merge(decile_lookup[decile_lookup["city"] == city],
                                            on="user_id", how="inner")
        # decent-mobility compliance threshold
        df["compliant"] = df["co2_kg"] <= 7  # 7 kg CO2e/week budget

        summary = (
            df.groupby("income_decile")["compliant"]
            .agg(["mean", "count"])
            .rename(columns={"mean": "compliance_rate"})
        )
        summary["city"] = city
        summary["mode"] = mode
        compliance_rows.append(summary.reset_index())

compliance_by_decile = pd.concat(compliance_rows, ignore_index=True)
print(compliance_by_decile.head(20))
```

```python
compliance_rows = []
for city in cities:
    for mode in modes:
        df = data_dict[(city, mode)].merge(
            decile_lookup[decile_lookup["city"] == city],
            on="user_id", how="inner"
        )
        df["compliant"] = df["co2_kg"] <= 7  # 7 kg CO2e/week budget

        summary = (
            df.groupby("income_decile")
            .agg(compliance_rate=("compliant", "mean"),
                 avg_co2_kg=("co2_kg", "mean"),
                 n=("compliant", "count"))
        )
        summary["city"] = city
        summary["mode"] = mode
        compliance_rows.append(summary.reset_index())

compliance_by_decile = pd.concat(compliance_rows, ignore_index=True)
```

```python
import numpy as np
def concentration_index(df, outcome_col, income_col):
    d = df.dropna(subset=[outcome_col, income_col]).copy()
    d["rank"] = d[income_col].rank(method="average", pct=True)
    y = d[outcome_col].astype(float)
    y_mean = y.mean()
    cov = np.cov(y, d["rank"], bias=True)[0, 1]
    ci = (2 / y_mean) * cov
    return ci, len(d)

ci_rows = []
for city in cities:
    for mode in modes:
        df = data_dict[(city, mode)].merge(
            gdf_home_income_final[gdf_home_income_final["city"] == city][["user_id", "hr_mtu"]],
            on="user_id", how="inner"
        )

        n_before = len(df)
        df = df.dropna(subset=["co2_kg", "hr_mtu"])
        n_after = len(df)
        if n_before != n_after:
            print(f"{city} {mode}: dropped {n_before - n_after} rows with missing co2_kg/tr_mtu")

        df["compliant"] = (df["co2_kg"] <= 7).astype(int)  # now safe, no NA left

        ci_val, n = concentration_index(df, "compliant", "hr_mtu")
        ci_rows.append({"city": city, "mode": mode, "concentration_index": ci_val, "n": n})

ci_table = pd.DataFrame(ci_rows)
print(ci_table)
```

```python
def concentration_curve_data(df, outcome_col, income_col):
    d = df.dropna(subset=[outcome_col, income_col]).copy()
    d = d.sort_values(income_col)  # poorest to richest
    d["cum_pop_share"] = np.arange(1, len(d) + 1) / len(d)
    d["cum_outcome_share"] = d[outcome_col].cumsum() / d[outcome_col].sum()
    return d[["cum_pop_share", "cum_outcome_share"]]

pt_curves = {}
for city in cities:
    df = data_dict[(city, "PT")].merge(
        gdf_home_income_final[gdf_home_income_final["city"] == city][["user_id", "hr_mtu"]],
        on="user_id", how="inner"
    )
    df = df.dropna(subset=["co2_kg", "hr_mtu"])
    df["compliant"] = (df["co2_kg"] <= 7).astype(int)

    pt_curves[city] = concentration_curve_data(df, "compliant", "hr_mtu")
```

```python
def bivariate_terciles(gdf_home, data_dict, city, income_col="hr_mtu"):
    df_co2 = data_dict[(city, "PT")][["user_id", "co2_kg"]]
    sub = gdf_home[gdf_home["city"] == city].merge(df_co2, on="user_id", how="inner")
    sub = sub.dropna(subset=[income_col, "co2_kg"])

    sub["income_tercile"] = pd.qcut(sub[income_col], 3, labels=[1, 2, 3]).astype(int)
    sub["co2_tercile"] = pd.qcut(sub["co2_kg"], 3, labels=[1, 2, 3]).astype(int)

    return sub

bivariate_data = {}
for city in cities:
    bivariate_data[city] = bivariate_terciles(gdf_home_income_final, data_dict, city, income_col="hr_mtu")
```

```python
# Rows = income tercile (1=low, 3=high), Cols = co2 tercile (1=low, 3=high)
# Standard sequential bivariate palette (light = both low, dark = both high)
bivariate_colors = {
    (1, 1): "#e8e8e8", (1, 2): "#b0d5df", (1, 3): "#64acbe",
    (2, 1): "#e4acac", (2, 2): "#ad9ea5", (2, 3): "#627f8c",
    (3, 1): "#c85a5a", (3, 2): "#985356", (3, 3): "#574249",
}
gdf_hex_by_city = {}
for city in cities:
    df = bivariate_data[city]
    df["biv_color"] = df.apply(
        lambda r: bivariate_colors[(r["income_tercile"], r["co2_tercile"])], axis=1
    )
```

```python
import matplotlib.colors as mcolors
import numpy as np
def build_bivariate_palette_v3():
    # Base hues for income tercile: red (low/poor) -> purple (mid) -> green (high)
    income_base = {
        1: np.array(mcolors.to_rgb("#c0442c")),  # burnt red — low income
        2: np.array(mcolors.to_rgb("#5b3d8a")),  # purple — mid income
        3: np.array(mcolors.to_rgb("#1b7a72")),  # teal-green — high income
    }
    co2_lighten = {1: 0.55, 2: 0.28, 3: 0.0}

    palette = {}
    for income_t, base_rgb in income_base.items():
        for co2_t, lighten in co2_lighten.items():
            mixed = base_rgb * (1 - lighten) + np.array([1, 1, 1]) * lighten
            palette[(income_t, co2_t)] = mcolors.to_hex(mixed)
    return palette

bivariate_colors = build_bivariate_palette_v3()

# Re-apply to existing hex frames
for city in cities:
    gdf_hex_by_city[city]["biv_color"] = gdf_hex_by_city[city].apply(
        lambda r: bivariate_colors[(r["income_tercile"], r["co2_tercile"])], axis=1
    )
gdf_hex_by_city = {}

for city in cities:
    plot_bivariate_map(city, bivariate_data[city], f"./output/{city.lower()}_bivariate_income_co2.png")
```
