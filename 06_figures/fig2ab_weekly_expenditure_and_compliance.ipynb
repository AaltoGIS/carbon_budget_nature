```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =====================================================
# Load data for all cities and modes
# =====================================================

cities = ["Helsinki", "Turku", "Tampere", "Oulu"]
modes = ["Cycling", "PT", "Car"]

data_dict = {}

for city in cities:
    for mode in modes:
        if mode == "Cycling":
            df = pd.read_parquet(
                f"./output/tour_bike_{city.lower()}_decent_per_user.parquet"
            )
        elif mode == "PT":
            df = pd.read_parquet(
                f"./output/tour_pt_{city.lower()}_decent_per_user.parquet"
            )
        elif mode == "Car":
            df = pd.read_parquet(
                f"./output/tour_car_{city.lower()}_decent_per_user.parquet"
            )
        
        data_dict[(city, mode)] = df["decent_mobility_co2"] / 1000  # Convert to kg

# =====================================================
# Settings
# =====================================================

BUDGET_KG = 7  # kg CO₂e/week

mode_colors = {
    "Cycling": "#2ecc71",      # Green
    "PT": "#9b59b6",           # Purple
    "Car": "#e74c3c"           # Red
}

# =====================================================
# Create 3x4 grid (3 modes × 4 cities) - SAME Y SCALE
# =====================================================

fig, axes = plt.subplots(
    nrows=3,
    ncols=4,
    figsize=(16, 10),
    sharey="row"  # SAME y-axis within each mode (row)
)

for row, mode in enumerate(modes):
    for col, city in enumerate(cities):
        
        ax = axes[row, col]
        
        # Get data
        co2_values = data_dict[(city, mode)]
        
        # Violin plot
        parts = ax.violinplot(
            [co2_values],
            positions=[0],
            widths=0.5,
            showmeans=False,
            showmedians=False,
            showextrema=False
        )
        
        # Color the violin
        for pc in parts["bodies"]:
            pc.set_facecolor(mode_colors[mode])
            pc.set_alpha(0.65)
            pc.set_edgecolor("black")
            pc.set_linewidth(1.2)
        
        # Scatter points with jitter
        jittered_x = np.random.normal(0, 0.035, size=len(co2_values))
        ax.scatter(
            jittered_x,
            co2_values,
            alpha=0.25,
            s=20,
            color=mode_colors[mode],
            edgecolor="none"
        )
        
        # Budget line
        ax.axhline(
            BUDGET_KG,
            color="#d62728",
            linestyle="--",
            linewidth=2.5,
            alpha=0.8
        )
        
        # Formatting
        ax.set_xlim(-0.6, 0.6)
        ax.set_xticks([])
        
        # Y-axis label (left column only)
        if col == 0:
            ax.set_ylabel(
                f"{mode}\nCO₂ (kg/week)",
                fontsize=11,
                fontweight="bold",
                color=mode_colors[mode]
            )
        
        # Title (top row only)
        if row == 0:
            ax.set_title(
                city,
                fontsize=12,
                fontweight="bold"
            )
        
        # Grid
        ax.grid(axis="y", linestyle=":", alpha=0.3, linewidth=0.8)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        ax.spines["bottom"].set_visible(False)
        ax.spines["left"].set_linewidth(1.2)

# =====================================================
# Global formatting
# =====================================================

fig.suptitle(
    "Weekly CO₂ Expenditure: Distribution by Transport Mode and City",
    fontsize=15,
    fontweight="bold",
    y=0.995
)

# Add legend for budget line
fig.text(
    0.5,
    0.02,
    "Red dashed line = 7 kg CO₂e/week target",
    ha="center",
    fontsize=11,
    style="italic"
)

plt.tight_layout(rect=[0, 0.035, 1, 0.985])
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# Load data for all cities and modes
# =====================================================

cities = ["Helsinki", "Turku", "Tampere", "Oulu"]
modes = ["Cycling", "PT", "Car"]

data_list = []
labels = []
colors = []
positions = []

mode_colors = {
    "Cycling": "#2ecc71",      # Green
    "PT": "#9b59b6",           # Purple
    "Car": "#e74c3c"           # Red
}

x_pos = 0

for city in cities:
    for mode in modes:
        if mode == "Cycling":
            df = pd.read_parquet(
                f"./output/tour_bike_{city.lower()}_decent_per_user.parquet"
            )
        elif mode == "PT":
            df = pd.read_parquet(
                f"./output/tour_pt_{city.lower()}_decent_per_user.parquet"
            )
        elif mode == "Car":
            df = pd.read_parquet(
                f"./output/tour_car_{city.lower()}_decent_per_user.parquet"
            )
        
        co2_values = df["decent_mobility_co2"] / 1000  # Convert to kg
        
        data_list.append(co2_values)
        labels.append(f"{mode}")
        colors.append(mode_colors[mode])
        positions.append(x_pos)
        
        x_pos += 1
    
    x_pos += 0.5  # Space between cities

# =====================================================
# Settings
# =====================================================

BUDGET_KG = 7  # kg CO₂e/week

# =====================================================
# Create single plot - ONE Y-AXIS FOR ALL (BOXPLOTS)
# =====================================================

fig, ax = plt.subplots(figsize=(18, 7))

# Boxplot
bp = ax.boxplot(
    data_list,
    positions=positions,
    widths=0.6,
    patch_artist=True,
    showfliers=False,
    medianprops=dict(color="black", linewidth=2.5),
    boxprops=dict(linewidth=1.5),
    whiskerprops=dict(linewidth=1.5),
    capprops=dict(linewidth=1.5)
)

# Color the boxes
for patch, color in zip(bp["boxes"], colors):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
    patch.set_edgecolor("black")

# =====================================================
# Budget line
# =====================================================

ax.axhline(
    BUDGET_KG,
    color="#d62728",
    linestyle="--",
    linewidth=3,
    alpha=0.8,
    label="7 kg CO₂e/week target"
)

# =====================================================
# X-axis labels (cities)
# =====================================================

city_positions = []
for i, city in enumerate(cities):
    city_pos = np.mean([positions[i*3], positions[i*3+2]])
    city_positions.append(city_pos)

ax.set_xticks(city_positions)
ax.set_xticklabels(cities, fontsize=13, fontweight="bold")

# =====================================================
# Mode labels below each city
# =====================================================

y_label_pos = ax.get_ylim()[0] - (ax.get_ylim()[1] - ax.get_ylim()[0]) * 0.05

for i, city_idx in enumerate(range(len(cities))):
    base_pos = city_idx * 3.5
    
    ax.text(
        base_pos,
        y_label_pos,
        "Cycling",
        ha="center",
        fontsize=11,
        color="#2ecc71",
        fontweight="bold"
    )
    
    ax.text(
        base_pos + 1,
        y_label_pos,
        "PT",
        ha="center",
        fontsize=11,
        color="#9b59b6",
        fontweight="bold"
    )
    
    ax.text(
        base_pos + 2,
        y_label_pos,
        "Car",
        ha="center",
        fontsize=11,
        color="#e74c3c",
        fontweight="bold"
    )

# =====================================================
# Formatting
# =====================================================

ax.set_ylabel(
    "Weekly CO₂ (kg/week)",
    fontsize=13,
    fontweight="bold"
)

ax.set_title(
    "Weekly CO₂ Expenditure: All Cities & Modes on Single Scale",
    fontsize=15,
    fontweight="bold",
    pad=20
)

ax.grid(axis="y", linestyle=":", alpha=0.4, linewidth=1)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.set_xlim(-0.8, x_pos - 0.5)

ax.legend(
    fontsize=11,
    loc="upper left",
    frameon=True,
    fancybox=False,
    edgecolor="black"
)

plt.subplots_adjust(bottom=0.12, left=0.08)
plt.show()
```

```python
"""
Panel A: ridgeline distributions of weekly CO2e by city and travel mode.

Nature-style aesthetics.

Each city contains three horizontal KDE distributions:
    Cycling
    PT
    Car

The distribution below the 7 kg CO2e target is shown with the main mode
colour, while the distribution above the target is shown with a lighter
version of the same colour.

A vertical white line marks the median of each distribution.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.lines import Line2D
from scipy.stats import gaussian_kde
import matplotlib.colors as mcolors
import matplotlib.patheffects as pe


# ---------------------------------------------------------------
# Clean data
# ---------------------------------------------------------------

plot_df = plot_df.copy()

plot_df["CO2"] = pd.to_numeric(
    plot_df["CO2"],
    errors="coerce"
)

plot_df = plot_df.dropna(
    subset=["CO2"]
)

plot_df["CO2"] = np.clip(
    plot_df["CO2"],
    0,
    None
)


# ---------------------------------------------------------------
# Order
# ---------------------------------------------------------------

cities = [
    "Oulu",
    "Tampere",
    "Turku",
    "Helsinki"
]

modes = [
    "Cycling",
    "PT",
    "Car"
]

tick_label_map = {
    "Cycling": "Cycling",
    "PT": "PT",
    "Car": "Car"
}


# ---------------------------------------------------------------
# Mode colours
# Same palette as panel B
# ---------------------------------------------------------------

mode_colors = {
    "Cycling": "#9FACCA",   # A2 - dusty blue-grey
    "PT": "#574E9C",        # B3 - deep indigo
    "Car": "#9A2588",       # C2 - muted plum-magenta
}


# ---------------------------------------------------------------
# Target
# ---------------------------------------------------------------

TARGET = 7.0


# ---------------------------------------------------------------
# Nature-style rcParams
# ---------------------------------------------------------------

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": [
        "Arial",
        "Helvetica",
        "DejaVu Sans"
    ],
    "font.size": 8,
    "axes.linewidth": 0.6,
    "xtick.major.width": 0.6,
    "ytick.major.width": 0.6,
    "xtick.major.size": 2.5,
    "ytick.major.size": 2.5,
    "svg.fonttype": "none",
    "pdf.fonttype": 42,
})


# ---------------------------------------------------------------
# Helper: lighter version of colour
# ---------------------------------------------------------------

def lighten_colour(
    colour,
    amount=0.72
):

    rgb = np.array(
        mcolors.to_rgb(colour)
    )

    white = np.ones(3)

    return tuple(
        rgb * (1 - amount)
        + white * amount
    )


# ---------------------------------------------------------------
# KDE helper
# ---------------------------------------------------------------

def get_kde(
    values,
    x_grid
):

    values = np.asarray(
        values,
        dtype=float
    )

    values = values[
        np.isfinite(values)
    ]

    if len(values) < 3:

        return np.zeros_like(
            x_grid
        )

    # Avoid KDE problems when all values are identical
    if np.ptp(values) == 0:

        sd = max(
            values.mean() * 0.03,
            0.05
        )

        density = np.exp(
            -0.5
            * (
                (x_grid - values.mean())
                / sd
            ) ** 2
        )

    else:

        kde = gaussian_kde(
            values,
            bw_method="scott"
        )

        density = kde(
            x_grid
        )

    # Normalize each ridge
    if density.max() > 0:

        density = (
            density
            / density.max()
        )

    return density


# ---------------------------------------------------------------
# Figure
# ---------------------------------------------------------------

fig, ax = plt.subplots(
    figsize=(7.2, 6.0),
    dpi=300
)


# ---------------------------------------------------------------
# X grid
# ---------------------------------------------------------------

upper = plot_df["CO2"].quantile(
    0.995
)

x_max = upper * 1.08

x_grid = np.linspace(
    0,
    x_max,
    700
)


# ---------------------------------------------------------------
# Ridge geometry
# ---------------------------------------------------------------

ridge_height = 0.72

mode_offsets = {
    "Cycling": 1.90,
    "PT": 0.95,
    "Car": 0.00
}

city_gap = 0.65


# ---------------------------------------------------------------
# Draw ridges
# ---------------------------------------------------------------

city_centers = []

for city_idx, city in enumerate(cities):

    city_base = (
        city_idx
        * (
            max(mode_offsets.values())
            + 1
            + city_gap
        )
    )

    city_positions = []

    for mode in modes:

        values = (
            plot_df.loc[
                (plot_df["City"] == city)
                &
                (plot_df["Mode"] == mode),
                "CO2"
            ]
            .astype(float)
            .values
        )

        values = values[
            np.isfinite(values)
        ]

        if len(values) == 0:
            continue

        density = get_kde(
            values,
            x_grid
        )

        y = (
            city_base
            + mode_offsets[mode]
        )

        city_positions.append(
            y
        )

        colour = mode_colors[mode]

        light_colour = lighten_colour(
            colour,
            amount=0.72
        )


        # -------------------------------------------------------
        # Distribution below target
        # -------------------------------------------------------

        mask_below = (
            x_grid <= TARGET
        )

        ax.fill_between(
            x_grid[mask_below],
            y,
            y
            + ridge_height
            * density[mask_below],
            color=colour,
            alpha=0.90,
            linewidth=0,
            zorder=2
        )


        # -------------------------------------------------------
        # Distribution above target
        # -------------------------------------------------------

        mask_above = (
            x_grid >= TARGET
        )

        ax.fill_between(
            x_grid[mask_above],
            y,
            y
            + ridge_height
            * density[mask_above],
            color=light_colour,
            alpha=0.85,
            linewidth=0,
            zorder=1
        )


        # -------------------------------------------------------
        # Thin white ridge outline
        # -------------------------------------------------------

        ax.plot(
            x_grid,
            y
            + ridge_height
            * density,
            color="white",
            linewidth=0.8,
            alpha=0.85,
            zorder=4
        )


        # -------------------------------------------------------
        # Median marker
        # Vertical line slicing through the distribution
        # -------------------------------------------------------

        median = np.median(
            values
        )

        median_density = np.interp(
            median,
            x_grid,
            density
        )

        # Vertical line from the baseline to the KDE
        ax.plot(
            [
                median,
                median
            ],
            [
                y,
                y
                + ridge_height
                * median_density
            ],
            color="white",
            linewidth=2,
            solid_capstyle="butt",
            zorder=7,
            path_effects=[
                pe.withStroke(
                    linewidth=2.0,
                    foreground="#555555"
                )
            ]
        )


        # -------------------------------------------------------
        # Mode label
        # -------------------------------------------------------

        ax.text(
            -0.015,
            y
            + ridge_height * 0.35,
            tick_label_map[mode],
            transform=ax.get_yaxis_transform(),
            ha="right",
            va="center",
            fontsize=8.5,
            color="#333333"
        )


    # -----------------------------------------------------------
    # City centre
    # -----------------------------------------------------------

    if city_positions:

        city_centers.append(
            np.mean(city_positions)
        )


# ---------------------------------------------------------------
# 7 kg target line
# ---------------------------------------------------------------

ax.axvline(
    TARGET,
    color="#333333",
    linestyle=(0, (4, 3)),
    linewidth=1.1,
    zorder=5
)




# ---------------------------------------------------------------
# City labels
# ---------------------------------------------------------------

for city, center in zip(
    cities,
    city_centers
):

    ax.text(
        -0.13,
        center,
        city,
        transform=ax.get_yaxis_transform(),
        ha="right",
        va="center",
        fontsize=9,
        fontweight="bold",
        color="#111111"
    )


# ---------------------------------------------------------------
# Axes
# ---------------------------------------------------------------

ax.set_xlim(
    0,
    x_max
)

ax.set_ylim(
    -0.35,
    max(city_centers) + 2.7
)

ax.set_yticks([])

ax.set_xlabel(
    "CO$_2$e (kg/week per person)",
    fontsize=9,
    labelpad=5
)

ax.xaxis.grid(
    True,
    color="#E5E5E5",
    linewidth=0.6,
    zorder=0
)

ax.set_axisbelow(
    True
)


# ---------------------------------------------------------------
# Spines
# ---------------------------------------------------------------

ax.spines["top"].set_visible(
    False
)

ax.spines["right"].set_visible(
    False
)

ax.spines["left"].set_visible(
    False
)

ax.spines["bottom"].set_color(
    "#444444"
)

ax.spines["bottom"].set_linewidth(
    0.6
)

ax.tick_params(
    axis="y",
    length=0
)

ax.tick_params(
    axis="x",
    color="#444444"
)


# ---------------------------------------------------------------
# Legend
# ---------------------------------------------------------------

handles = [
    mpatches.Patch(
        facecolor=mode_colors[m],
        edgecolor="none",
        alpha=0.9,
        label=tick_label_map[m]
    )
    for m in modes
]

handles.append(
    Line2D(
        [0],
        [0],
        color="#333333",
        linestyle=(0, (4, 3)),
        lw=1.1,
        label="7 kg CO$_2$e target"
    )
)

legend = ax.legend(
    handles=handles,
    loc="upper center",
    bbox_to_anchor=(0.5, -0.08),
    ncol=4,
    frameon=False,
    fontsize=8.5,
    handlelength=1.2,
    columnspacing=1.3
)


# ---------------------------------------------------------------
# Save
# ---------------------------------------------------------------

plt.tight_layout()

out_png = (
    "./output/"
    "co2_ridgeline_panelA.png"
)

out_svg = (
    "./output/"
    "co2_ridgeline_panelA.svg"
)

plt.savefig(
    out_png,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.savefig(
    out_svg,
    bbox_inches="tight",
    facecolor="white"
)

print(
    "Saved:",
    out_png,
    out_svg
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# -----------------------------
# Clean data
# -----------------------------
plot_df = plot_df.copy()
plot_df["CO2"] = pd.to_numeric(plot_df["CO2"], errors="coerce")
plot_df = plot_df.dropna(subset=["CO2"])

# -----------------------------
# Order
# -----------------------------
cities = ["Helsinki", "Turku", "Tampere", "Oulu"]
modes = ["Cycling", "PT", "Car"]

mode_colors = {
    "Cycling": "#FFFFFF",
    "PT": "#C0C0C0",
    "Car": "#4D4D4D"
}

# -----------------------------
# Build data
# -----------------------------
box_data = []
positions = []
labels = []
colors = []
city_centers = []

gap = 1.6   # more breathing space
y = 0.1

for city in cities:

    city_positions = []

    for mode in modes:

        values = (
            plot_df.loc[
                (plot_df["City"] == city) &
                (plot_df["Mode"] == mode),
                "CO2"
            ]
            .astype(float)
            .values
        )

        # IMPORTANT FIX: no negative emissions in plot
        values = np.clip(values, 0, None)

        box_data.append(values)
        positions.append(y)
        labels.append(mode)
        colors.append(mode_colors[mode])

        city_positions.append(y)
        y += 1

    city_centers.append(np.mean(city_positions))
    y += gap

# -----------------------------
# Figure (bigger!)
# -----------------------------
fig, ax = plt.subplots(figsize=(12, 7))

bp = ax.boxplot(
    box_data,
    positions=positions,
    vert=False,
    widths=0.8,              # FATTER BOXES
    patch_artist=True,
    showfliers=False,
)

# -----------------------------
# Styling (clean + readable)
# -----------------------------
for box, c in zip(bp["boxes"], colors):
    box.set_facecolor(c)
    box.set_edgecolor("black")
    box.set_linewidth(1.2)

for whisker in bp["whiskers"]:
    whisker.set_color("black")
    whisker.set_linewidth(0.8)

for cap in bp["caps"]:
    cap.set_color("black")
    cap.set_linewidth(0.8)

for median in bp["medians"]:
    median.set_color("black")
    median.set_linewidth(2.5)   # emphasize median

# -----------------------------
# 7 kg target line
# -----------------------------
ax.axvline(
    7,
    color="blue",
    linestyle="--",
    linewidth=1.5,
    label="7 kg target",
)

# -----------------------------
# Labels
# -----------------------------
ax.set_yticks(positions)
ax.set_yticklabels(labels, fontsize=12)

for city, c in zip(cities, city_centers):
    ax.text(
        -0.06,
        c,
        city,
        transform=ax.get_yaxis_transform(),
        ha="right",
        va="center",
        fontsize=13,
        fontweight="bold"
    )

# -----------------------------
# Layout
# -----------------------------
ax.set_xlabel("CO₂ (kg/week per person)", fontsize=13)
ax.set_ylabel("")

upper = plot_df["CO2"].quantile(0.995)
ax.set_xlim(0, upper * 1.1)

ax.grid(axis="x", color="0.9", linewidth=1)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.invert_yaxis()

# -----------------------------
# Legend
# -----------------------------
handles = [
    mpatches.Patch(facecolor=mode_colors[m], edgecolor="black", label=m)
    for m in modes
]

handles.append(plt.Line2D([0], [0], color="blue", linestyle="--", label="Target:\n7 kg CO$_2$"))

ax.legend(handles=handles, frameon=False)

plt.tight_layout()

plt.savefig(
    "./output/co2_boxplot_panelA.png",
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)
plt.show()
```

```python
import pandas as pd
import numpy as np

# ensure clean
df = plot_df.copy()
df["CO2"] = pd.to_numeric(df["CO2"], errors="coerce")
df = df.dropna(subset=["CO2"])

# threshold
THRESHOLD = 7

rows = []

for city in cities:
    for mode in modes:

        subset = df[(df["City"] == city) & (df["Mode"] == mode)]

        if len(subset) == 0:
            pct = np.nan
        else:
            pct = (subset["CO2"] <= THRESHOLD).mean() * 100

        rows.append({
            "City": city,
            "Mode": mode,
            "% ≤ 7 kg CO₂e": round(pct, 1)
        })

panel_b = pd.DataFrame(rows)
```
